```python
#Cmkt
def cmkt(df):
    da=pd.DataFrame()
    da=df[['market_cap','canonical_id']].copy()
    da['market_returns']=da['market_cap'].pct_change()
    #da['riskfreeval']=(1+(apy/100))**(1/365)-1
    da['cmkt']=da['market_returns'].apply(excessreturns)
    return da


dataset=cmkt(df)
df=pd.concat([df,dataset[['market_returns','cmkt']]],axis=1)
df
```

```python
#value = (TVL / MarketCap)
#hml_signal = 1 / value
def hml(df):
    de=pd.DataFrame()
    de=df[['market_cap','canonical_id','protocol_tvl']].copy()
    de['market_cap']
    de['value']=de['protocol_tvl']/de['market_cap']
    de['hml']=1/de['value']
    return de

de=hml(df)
df = pd.concat([df,de[['value','hml']]],axis=1)
```

```python
#calculate smb
def smb(df):
    dea=df.copy()
    dea['returns']=df['close'].pct_change()
    dea=df.sort_values('market_cap',ascending=True)
    dea['returns']=df['close'].pct_change()
    numtokens=len(dea)
    val=int(numtokens*0.3)
    deas=dea.head(val)
    deal=dea.tail(val)
    dea['smb']=deas['returns'].mean()-deal['returns'].mean()
    return dea
dea=smb(df)
df=pd.concat([df,dea['smb']],axis=1)
```

```python
def calculate_altcoin_dominance(df):
    daily_top10 = df.groupby('date').apply(
        lambda x: x.nlargest(10, 'market_cap')['market_cap'].sum())
    total_mcap = df.groupby('date')['market_cap'].sum()
    altcoin_dominance = ((total_mcap - daily_top10) / total_mcap) * 100
    return df['date'].map(altcoin_dominance)
```
